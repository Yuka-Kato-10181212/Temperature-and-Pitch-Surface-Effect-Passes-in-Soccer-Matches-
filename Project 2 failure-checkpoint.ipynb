```python
#create a function to keep the necessary columns and filter the 'type' column to keep the rows 'Pass' + reset the index

def cleaning(input_df):

    necessary_columns = [
        'location', 'pass_angle', 'pass_end_location', 'pass_length', 'pass_outcome', 'pass_switch', 'period',
        'play_pattern', 'player', 'position', 'possession','possession_team', 'possession_team_id', 'team',
        'team_id', 'timestamp', 'type', 'duration', 'period', 'win_or_lose', 'home_or_away', 'category', 'sample_num'
    ]

    cleaned_df = input_df[necessary_columns]

    cleaned_df = cleaned_df[cleaned_df['type'] == 'Pass']
    
    cleaned_df['num_pass_outcome'] = [0 if x == 'Incomplete' else 1 for x in cleaned_df['pass_outcome']]

    cleaned_df.reset_index(drop=True, inplace=True)

    return cleaned_df
```
